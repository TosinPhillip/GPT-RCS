```python
import bcrypt

# Plain-text password provided by the user (e.g., from a login form)
plain_password = "123456"

# Hashed password retrieved from the database (stored as a string)
stored_hash = "$2b$12$mSbQlYj1Kt10B0zFjsJbHuB2CxdJwwvuv2FP0rbQMpyrndaQK.Yaq"

# Convert both values to bytes (required by bcrypt)
plain_password_bytes = plain_password.encode('utf-8')
stored_hash_bytes = stored_hash.encode('utf-8')

# Compare passwords
if bcrypt.checkpw(plain_password_bytes, stored_hash_bytes):
    print("Password is correct")
else:
    print("Invalid password")

```
