```python
# Fill all the passwords with 123456
std.password = std['password'].fillna('123456')

std['password']
```

```python
def hash_password(plain_password):
    # convert password to bytes
    plain_password_bytes = plain_password.encode('utf-8')
    
    # generate salt
    salt = bcrypt.gensalt()
    
    # hash password
    hashed_bytes = bcrypt.hashpw(plain_password_bytes, salt)
    
    # store as string
    stored_hash = hashed_bytes.decode('utf-8')
    
    return stored_hash
```

```python
hash_password('123456')
```

```python
std['password'] = std['password'].astype('str').apply(hash_password)

std['password']
```
